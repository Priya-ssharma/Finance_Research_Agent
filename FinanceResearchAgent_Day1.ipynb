```python

# FinanceResearchAgent — Step 8
# Reusable, source-linked research search tool

import json
import re
from pathlib import Path
from IPython.display import display, Markdown

DATA_PATH = Path(
    "/content/finance_research_output/ril_annual_report_pages.json"
)

with open(DATA_PATH, "r", encoding="utf-8") as f:
    report_pages = json.load(f)


def research_search(question, max_results=5):
    """
    Search extracted annual-report pages for relevant evidence.
    Returns source-linked passages, not AI-generated conclusions.
    """
    if not question or not question.strip():
        raise ValueError("Please enter a research question.")

    if not isinstance(max_results, int) or max_results < 1:
        raise ValueError("max_results must be a positive integer.")

    # Remove common question words and score pages by matching terms.
    stop_words = {
        "what", "which", "when", "where", "why", "how",
        "does", "did", "the", "and", "for", "from",
        "with", "that", "this", "are", "was", "were",
        "explain", "tell", "about", "reliance", "industries"
    }

    terms = [
        word.lower()
        for word in re.findall(r"[A-Za-z0-9]+", question)
        if word.lower() not in stop_words and len(word) > 2
    ]

    if not terms:
        raise ValueError("Please include a specific research topic.")

    results = []

    for page in report_pages:
        text = re.sub(r"\s+", " ", page["text"]).strip()
        lower_text = text.lower()

        score = sum(lower_text.count(term) for term in terms)

        if score == 0:
            continue

        # Find the first matching passage.
        first_match = min(
            (lower_text.find(term) for term in terms
             if lower_text.find(term) >= 0),
            default=0
        )

        start = max(0, first_match - 250)
        end = min(len(text), first_match + 650)

        results.append({
            "score": score,
            "question": question,
            "pdf_page": page["pdf_page"],
            "reporting_period": page["reporting_period"],
            "source_file": page["source_file"],
            "source_url": page["source_url"],
            "evidence": text[start:end]
        })

    results.sort(key=lambda item: item["score"], reverse=True)
    return results[:max_results]


# Test the tool
question = "What explains the change in net profit margin?"

results = research_search(question, max_results=5)

print("Research question:", question)
print("Matching passages:", len(results))

for i, item in enumerate(results, start=1):
    print(f"\n--- Result {i} | PDF page {item['pdf_page']} ---")
    print("Reporting period:", item["reporting_period"])
    print("Source:", item["source_url"])
    print("Evidence:", item["evidence"])
```

Step 9 — Improve the research search function

```python

# FinanceResearchAgent — Step 9
# Improved keyword-based evidence retrieval

import json
import re
from pathlib import Path

DATA_PATH = Path(
    "/content/finance_research_output/ril_annual_report_pages.json"
)

with open(DATA_PATH, "r", encoding="utf-8") as f:
    report_pages = json.load(f)


def research_search_v2(question, max_results=5):
    if not question or not question.strip():
        raise ValueError("Enter a research question.")

    q = question.lower()

    # Topic-specific terms and phrases
    topic_rules = {
        "net profit margin": [
            "net profit margin",
            "higher net profit",
            "8.0%",
            "6.3%"
        ],
        "debt service coverage ratio": [
            "debt service coverage ratio",
            "finance cost",
            "principal repayments"
        ],
        "return on capital employed": [
            "return on capital employed",
            "20.7%",
            "14.6%"
        ]
    }

    selected_topic = next(
        (topic for topic in topic_rules if topic in q),
        None
    )

    if selected_topic is None:
        terms = [
            t for t in re.findall(r"[a-z0-9]+", q)
            if len(t) > 2
        ]
    else:
        terms = topic_rules[selected_topic]

    results = []

    for page in report_pages:
        text = re.sub(r"\s+", " ", page["text"]).strip()
        lower_text = text.lower()

        score = 0

        for term in terms:
            if term in lower_text:
                score += 5 if " " in term else 1

        # Prefer the standalone ratio-analysis page for
        # questions about reported standalone financial ratios.
        if selected_topic and "ratio analysis" in lower_text:
            score += 10

        if score == 0:
            continue

        results.append({
            "score": score,
            "pdf_page": page["pdf_page"],
            "source_url": page["source_url"],
            "evidence": text[:1800]
        })

    results.sort(
        key=lambda x: (x["score"], -x["pdf_page"]),
        reverse=True
    )

    return results[:max_results]


question = "What explains the change in net profit margin?"
results = research_search_v2(question, max_results=3)

print("Question:", question)
print("Relevant pages found:", len(results))

for result in results:
    print(f"\nPDF page: {result['pdf_page']}")
    print("Relevance score:", result["score"])
    print("Source:", result["source_url"])
    print("Evidence:\n", result["evidence"])
```

```python

import re

def extract_relevant_sentences(question, evidence, max_sentences=4):
    """Extract source sentences that best match a research question."""

    question_terms = set(
        re.findall(r"[a-zA-Z]+", question.lower())
    )

    sentences = re.split(r"(?<=[.!?])\s+", evidence)
    scored_sentences = []

    for sentence in sentences:
        sentence_lower = sentence.lower()
        score = sum(
            1 for term in question_terms
            if term in sentence_lower
        )

        if score > 0:
            scored_sentences.append((score, sentence.strip()))

    scored_sentences.sort(key=lambda item: item[0], reverse=True)

    return [
        sentence
        for score, sentence in scored_sentences[:max_sentences]
    ]


question = "What explains the change in net profit margin?"
results = research_search_v2(question)

for result in results[:2]:
    print(f"\nPDF page: {result['pdf_page']}")

    relevant = extract_relevant_sentences(
        question,
        result["evidence"]
    )

    for sentence in relevant:
        print("-", sentence)
```

```python

def get_metric_evidence(question, results):
    question_lower = question.lower()

    metric_rules = {
        "net profit margin": [
            "net profit margin",
            "higher net profit"
        ],
        "debt service coverage ratio": [
            "debt service coverage ratio",
            "finance cost",
            "principal repayments"
        ],
        "return on capital employed": [
            "return on capital employed",
            "reduction in capital employed",
            "higher profit"
        ]
    }

    selected_metric = next(
        (
            metric for metric in metric_rules
            if metric in question_lower
        ),
        None
    )

    if selected_metric is None:
        return {
            "status": "needs_review",
            "message": "No metric-specific rule found.",
            "evidence": []
        }

    keywords = metric_rules[selected_metric]
    matches = []

    for result in results:
        text = result["evidence"]
        sentences = re.split(r"(?<=[.!?])\s+", text)

        for sentence in sentences:
            sentence_lower = sentence.lower()

            if (
                selected_metric in sentence_lower
                or (
                    "net profit margin" in selected_metric
                    and "net profit margin" in sentence_lower
                )
            ):
                matches.append({
                    "pdf_page": result["pdf_page"],
                    "source_url": result["source_url"],
                    "evidence": sentence.strip()
                })

    return {
        "status": "evidence_found" if matches else "needs_review",
        "metric": selected_metric,
        "evidence": matches
    }


question = "What explains the change in net profit margin?"
evidence_result = get_metric_evidence(question, results)

print("Status:", evidence_result["status"])
print("Metric:", evidence_result.get("metric", "Not identified"))

for item in evidence_result["evidence"]:
    print("\nPDF page:", item["pdf_page"])
    print("Evidence:", item["evidence"])
    print("Source:", item["source_url"])
```

extract only the relevant evidence

```python

def get_clean_metric_evidence(question, results):
    question_lower = question.lower()

    metric_rules = {
        "net profit margin": [
            "net profit margin increased from",
            "net profit margin increased due to"
        ],
        "debt service coverage ratio": [
            "debt service coverage ratio increased from",
            "debt service coverage ratio increased due to"
        ],
        "return on capital employed": [
            "return on capital employed increased from",
            "return on capital employed increased due to"
        ]
    }

    selected_metric = next(
        (metric for metric in metric_rules if metric in question_lower),
        None
    )

    if selected_metric is None:
        return {
            "status": "needs_review",
            "message": "Metric not supported by current rules.",
            "evidence": []
        }

    matches = []

    for result in results:
        text = result["evidence"]
        sentences = re.split(r"(?<=[.!?])\s+", text)

        for sentence in sentences:
            sentence_lower = sentence.lower()

            if any(
                phrase in sentence_lower
                for phrase in metric_rules[selected_metric]
            ):
                matches.append({
                    "pdf_page": result["pdf_page"],
                    "evidence": sentence.strip(),
                    "source_url": result["source_url"]
                })

    # Remove duplicate evidence while preserving source references.
    unique_matches = []
    seen = set()

    for match in matches:
        key = (match["pdf_page"], match["evidence"])

        if key not in seen:
            seen.add(key)
            unique_matches.append(match)

    return {
        "status": "evidence_found" if unique_matches else "needs_review",
        "metric": selected_metric,
        "evidence": unique_matches
    }


question = "What explains the change in net profit margin?"
clean_result = get_clean_metric_evidence(question, results)

print("Status:", clean_result["status"])
print("Metric:", clean_result.get("metric", "Not identified"))

for item in clean_result["evidence"]:
    print("\nPDF page:", item["pdf_page"])
    print("Evidence:", item["evidence"])
    print("Source:", item["source_url"])
```

build the research answer pipeline

```python

def run_finance_research(question):
    # Step 1: Retrieve relevant report pages
    search_results = research_search_v2(question)

    # Step 2: Extract evidence for supported metrics
    evidence_result = get_clean_metric_evidence(
        question,
        search_results
    )

    # Step 3: Check whether evidence is sufficient
    if evidence_result["status"] != "evidence_found":
        return {
            "question": question,
            "status": "needs_review",
            "answer": (
                "The current research rules could not find sufficient "
                "metric-specific evidence. Manual review is required."
            ),
            "sources": []
        }

    # Step 4: Generate a grounded answer for supported metrics
    metric = evidence_result["metric"]
    evidence = evidence_result["evidence"]

    if metric == "net profit margin":
        answer = (
            "Reliance Industries Limited's standalone net profit margin "
            "increased from 6.3% in FY 2024-25 to 8.0% in FY 2025-26, "
            "an increase of 1.7 percentage points. The annual report "
            "attributes the improvement to higher net profit."
        )
    elif metric == "debt service coverage ratio":
        answer = (
            "The annual report states that the debt service coverage "
            "ratio increased due to lower finance costs and principal "
            "repayments of loans during the year."
        )
    elif metric == "return on capital employed":
        answer = (
            "The annual report attributes the increase in return on "
            "capital employed to higher profit and a reduction in "
            "capital employed."
        )
    else:
        answer = "Manual review is required."

    # Step 5: Return answer and its supporting evidence
    return {
        "question": question,
        "status": "evidence_found",
        "answer": answer,
        "sources": evidence
    }


report = run_finance_research(
    "What explains the change in net profit margin?"
)

print("QUESTION:", report["question"])
print("\nSTATUS:", report["status"])
print("\nANSWER:", report["answer"])
print("\nSUPPORTING EVIDENCE:")

for source in report["sources"]:
    print(f"\nPDF page: {source['pdf_page']}")
    print("Evidence:", source["evidence"])
    print("Source:", source["source_url"])
```

make the agent more reliable

```python

def validate_research_evidence(metric, evidence_items):
    explanation_phrases = {
        "net profit margin": [
            "higher net profit"
        ],
        "debt service coverage ratio": [
            "lower finance cost",
            "principal repayments"
        ],
        "return on capital employed": [
            "higher profit",
            "reduction in capital employed"
        ]
    }

    metric = metric.lower()

    if metric not in explanation_phrases:
        return {
            "status": "needs_review",
            "reason": "No validation rule exists for this metric."
        }

    relevant_items = []

    for item in evidence_items:
        text = item["evidence"].lower()

        has_metric = metric in text
        has_explanation = any(
            phrase in text
            for phrase in explanation_phrases[metric]
        )

        if has_metric and has_explanation:
            relevant_items.append(item)

    return {
        "status": (
            "validated" if relevant_items else "needs_review"
        ),
        "metric": metric,
        "evidence": relevant_items,
        "reason": (
            "Metric and explanatory wording found in source text."
            if relevant_items
            else "Supporting explanation was not found."
        )
    }


validation = validate_research_evidence(
    "return on capital employed",
    test_report["sources"]
)

print("Validation status:", validation["status"])
print("Reason:", validation["reason"])

for item in validation.get("evidence", []):
    print("\nPDF page:", item["pdf_page"])
    print("Evidence:", item["evidence"])
```
