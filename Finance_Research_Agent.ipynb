Step 8 — Build your first reusable research tool

```python

# FinanceResearchAgent — Step 8
# Reusable, source-linked research search tool

import json
import re
from pathlib import Path
from IPython.display import display, Markdown

DATA_PATH = Path(
    "/content/finance_research_output/ril_annual_report_pages.json"
)

with open(DATA_PATH, "r", encoding="utf-8") as f:
    report_pages = json.load(f)


def research_search(question, max_results=5):
    """
    Search extracted annual-report pages for relevant evidence.
    Returns source-linked passages, not AI-generated conclusions.
    """
    if not question or not question.strip():
        raise ValueError("Please enter a research question.")

    if not isinstance(max_results, int) or max_results < 1:
        raise ValueError("max_results must be a positive integer.")

    # Remove common question words and score pages by matching terms.
    stop_words = {
        "what", "which", "when", "where", "why", "how",
        "does", "did", "the", "and", "for", "from",
        "with", "that", "this", "are", "was", "were",
        "explain", "tell", "about", "reliance", "industries"
    }

    terms = [
        word.lower()
        for word in re.findall(r"[A-Za-z0-9]+", question)
        if word.lower() not in stop_words and len(word) > 2
    ]

    if not terms:
        raise ValueError("Please include a specific research topic.")

    results = []

    for page in report_pages:
        text = re.sub(r"\s+", " ", page["text"]).strip()
        lower_text = text.lower()

        score = sum(lower_text.count(term) for term in terms)

        if score == 0:
            continue

        # Find the first matching passage.
        first_match = min(
            (lower_text.find(term) for term in terms
             if lower_text.find(term) >= 0),
            default=0
        )

        start = max(0, first_match - 250)
        end = min(len(text), first_match + 650)

        results.append({
            "score": score,
            "question": question,
            "pdf_page": page["pdf_page"],
            "reporting_period": page["reporting_period"],
            "source_file": page["source_file"],
            "source_url": page["source_url"],
            "evidence": text[start:end]
        })

    results.sort(key=lambda item: item["score"], reverse=True)
    return results[:max_results]


# Test the tool
question = "What explains the change in net profit margin?"

results = research_search(question, max_results=5)

print("Research question:", question)
print("Matching passages:", len(results))

for i, item in enumerate(results, start=1):
    print(f"\n--- Result {i} | PDF page {item['pdf_page']} ---")
    print("Reporting period:", item["reporting_period"])
    print("Source:", item["source_url"])
    print("Evidence:", item["evidence"])
```

Step 9 — Improve the research search function

```python

# FinanceResearchAgent — Step 9
# Improved keyword-based evidence retrieval

import json
import re
from pathlib import Path

DATA_PATH = Path(
    "/content/finance_research_output/ril_annual_report_pages.json"
)

with open(DATA_PATH, "r", encoding="utf-8") as f:
    report_pages = json.load(f)


def research_search_v2(question, max_results=5):
    if not question or not question.strip():
        raise ValueError("Enter a research question.")

    q = question.lower()

    # Topic-specific terms and phrases
    topic_rules = {
        "net profit margin": [
            "net profit margin",
            "higher net profit",
            "8.0%",
            "6.3%"
        ],
        "debt service coverage ratio": [
            "debt service coverage ratio",
            "finance cost",
            "principal repayments"
        ],
        "return on capital employed": [
            "return on capital employed",
            "20.7%",
            "14.6%"
        ]
    }

    selected_topic = next(
        (topic for topic in topic_rules if topic in q),
        None
    )

    if selected_topic is None:
        terms = [
            t for t in re.findall(r"[a-z0-9]+", q)
            if len(t) > 2
        ]
    else:
        terms = topic_rules[selected_topic]

    results = []

    for page in report_pages:
        text = re.sub(r"\s+", " ", page["text"]).strip()
        lower_text = text.lower()

        score = 0

        for term in terms:
            if term in lower_text:
                score += 5 if " " in term else 1

        # Prefer the standalone ratio-analysis page for
        # questions about reported standalone financial ratios.
        if selected_topic and "ratio analysis" in lower_text:
            score += 10

        if score == 0:
            continue

        results.append({
            "score": score,
            "pdf_page": page["pdf_page"],
            "source_url": page["source_url"],
            "evidence": text[:1800]
        })

    results.sort(
        key=lambda x: (x["score"], -x["pdf_page"]),
        reverse=True
    )

    return results[:max_results]


question = "What explains the change in net profit margin?"
results = research_search_v2(question, max_results=3)

print("Question:", question)
print("Relevant pages found:", len(results))

for result in results:
    print(f"\nPDF page: {result['pdf_page']}")
    print("Relevance score:", result["score"])
    print("Source:", result["source_url"])
    print("Evidence:\n", result["evidence"])
```

```python

import re

def extract_relevant_sentences(question, evidence, max_sentences=4):
    """Extract source sentences that best match a research question."""

    question_terms = set(
        re.findall(r"[a-zA-Z]+", question.lower())
    )

    sentences = re.split(r"(?<=[.!?])\s+", evidence)
    scored_sentences = []

    for sentence in sentences:
        sentence_lower = sentence.lower()
        score = sum(
            1 for term in question_terms
            if term in sentence_lower
        )

        if score > 0:
            scored_sentences.append((score, sentence.strip()))

    scored_sentences.sort(key=lambda item: item[0], reverse=True)

    return [
        sentence
        for score, sentence in scored_sentences[:max_sentences]
    ]


question = "What explains the change in net profit margin?"
results = research_search_v2(question)

for result in results[:2]:
    print(f"\nPDF page: {result['pdf_page']}")

    relevant = extract_relevant_sentences(
        question,
        result["evidence"]
    )

    for sentence in relevant:
        print("-", sentence)
```

```python

def get_metric_evidence(question, results):
    question_lower = question.lower()

    metric_rules = {
        "net profit margin": [
            "net profit margin",
            "higher net profit"
        ],
        "debt service coverage ratio": [
            "debt service coverage ratio",
            "finance cost",
            "principal repayments"
        ],
        "return on capital employed": [
            "return on capital employed",
            "reduction in capital employed",
            "higher profit"
        ]
    }

    selected_metric = next(
        (
            metric for metric in metric_rules
            if metric in question_lower
        ),
        None
    )

    if selected_metric is None:
        return {
            "status": "needs_review",
            "message": "No metric-specific rule found.",
            "evidence": []
        }

    keywords = metric_rules[selected_metric]
    matches = []

    for result in results:
        text = result["evidence"]
        sentences = re.split(r"(?<=[.!?])\s+", text)

        for sentence in sentences:
            sentence_lower = sentence.lower()

            if (
                selected_metric in sentence_lower
                or (
                    "net profit margin" in selected_metric
                    and "net profit margin" in sentence_lower
                )
            ):
                matches.append({
                    "pdf_page": result["pdf_page"],
                    "source_url": result["source_url"],
                    "evidence": sentence.strip()
                })

    return {
        "status": "evidence_found" if matches else "needs_review",
        "metric": selected_metric,
        "evidence": matches
    }


question = "What explains the change in net profit margin?"
evidence_result = get_metric_evidence(question, results)

print("Status:", evidence_result["status"])
print("Metric:", evidence_result.get("metric", "Not identified"))

for item in evidence_result["evidence"]:
    print("\nPDF page:", item["pdf_page"])
    print("Evidence:", item["evidence"])
    print("Source:", item["source_url"])
```

extract only the relevant evidence

```python

def get_clean_metric_evidence(question, results):
    question_lower = question.lower()

    metric_rules = {
        "net profit margin": [
            "net profit margin increased from",
            "net profit margin increased due to"
        ],
        "debt service coverage ratio": [
            "debt service coverage ratio increased from",
            "debt service coverage ratio increased due to"
        ],
        "return on capital employed": [
            "return on capital employed increased from",
            "return on capital employed increased due to"
        ]
    }

    selected_metric = next(
        (metric for metric in metric_rules if metric in question_lower),
        None
    )

    if selected_metric is None:
        return {
            "status": "needs_review",
            "message": "Metric not supported by current rules.",
            "evidence": []
        }

    matches = []

    for result in results:
        text = result["evidence"]
        sentences = re.split(r"(?<=[.!?])\s+", text)

        for sentence in sentences:
            sentence_lower = sentence.lower()

            if any(
                phrase in sentence_lower
                for phrase in metric_rules[selected_metric]
            ):
                matches.append({
                    "pdf_page": result["pdf_page"],
                    "evidence": sentence.strip(),
                    "source_url": result["source_url"]
                })

    # Remove duplicate evidence while preserving source references.
    unique_matches = []
    seen = set()

    for match in matches:
        key = (match["pdf_page"], match["evidence"])

        if key not in seen:
            seen.add(key)
            unique_matches.append(match)

    return {
        "status": "evidence_found" if unique_matches else "needs_review",
        "metric": selected_metric,
        "evidence": unique_matches
    }


question = "What explains the change in net profit margin?"
clean_result = get_clean_metric_evidence(question, results)

print("Status:", clean_result["status"])
print("Metric:", clean_result.get("metric", "Not identified"))

for item in clean_result["evidence"]:
    print("\nPDF page:", item["pdf_page"])
    print("Evidence:", item["evidence"])
    print("Source:", item["source_url"])
```

build the research answer pipeline

```python

def run_finance_research(question):
    # Step 1: Retrieve relevant report pages
    search_results = research_search_v2(question)

    # Step 2: Extract evidence for supported metrics
    evidence_result = get_clean_metric_evidence(
        question,
        search_results
    )

    # Step 3: Check whether evidence is sufficient
    if evidence_result["status"] != "evidence_found":
        return {
            "question": question,
            "status": "needs_review",
            "answer": (
                "The current research rules could not find sufficient "
                "metric-specific evidence. Manual review is required."
            ),
            "sources": []
        }

    # Step 4: Generate a grounded answer for supported metrics
    metric = evidence_result["metric"]
    evidence = evidence_result["evidence"]

    if metric == "net profit margin":
        answer = (
            "Reliance Industries Limited's standalone net profit margin "
            "increased from 6.3% in FY 2024-25 to 8.0% in FY 2025-26, "
            "an increase of 1.7 percentage points. The annual report "
            "attributes the improvement to higher net profit."
        )
    elif metric == "debt service coverage ratio":
        answer = (
            "The annual report states that the debt service coverage "
            "ratio increased due to lower finance costs and principal "
            "repayments of loans during the year."
        )
    elif metric == "return on capital employed":
        answer = (
            "The annual report attributes the increase in return on "
            "capital employed to higher profit and a reduction in "
            "capital employed."
        )
    else:
        answer = "Manual review is required."

    # Step 5: Return answer and its supporting evidence
    return {
        "question": question,
        "status": "evidence_found",
        "answer": answer,
        "sources": evidence
    }


report = run_finance_research(
    "What explains the change in net profit margin?"
)

print("QUESTION:", report["question"])
print("\nSTATUS:", report["status"])
print("\nANSWER:", report["answer"])
print("\nSUPPORTING EVIDENCE:")

for source in report["sources"]:
    print(f"\nPDF page: {source['pdf_page']}")
    print("Evidence:", source["evidence"])
    print("Source:", source["source_url"])
```

```python

import re

def get_clean_metric_evidence(question, results):
    question_lower = question.lower()

    metric_rules = {
        "net profit margin": [
            "net profit margin increased from",
            "net profit margin increased due to",
            "net profit margin"
        ],
        "debt service coverage ratio": [
            "debt service coverage ratio increased from",
            "debt service coverage ratio increased due to",
            "debt service coverage ratio"
        ],
        "return on capital employed": [
            "return on capital employed increased from",
            "return on capital employed increased due to",
            "return on capital employed"
        ],
        "current ratio": ["current ratio"],
        "debt-equity ratio": ["debt-equity ratio", "debt equity ratio"],
        "return on equity": ["return on equity"],
        "inventory turnover ratio": ["inventory turnover ratio"],
        "trade receivables turnover ratio": [
            "trade receivables turnover ratio"
        ],
        "trade payables turnover ratio": [
            "trade payables turnover ratio"
        ],
        "return on investment": ["return on investment"],
        "interest coverage ratio": ["interest coverage ratio"],
        "long term debt to working capital": [
            "long term debt to working capital"
        ],
        "return on net worth": ["return on net worth"]
    }

    # Match the longest metric name first to avoid ambiguous matches.
    selected_metric = next(
        (
            metric for metric in sorted(
                metric_rules, key=len, reverse=True
            )
            if metric in question_lower
            or (
                metric == "debt-equity ratio"
                and "debt equity ratio" in question_lower
            )
        ),
        None
    )

    if selected_metric is None:
        return {
            "status": "needs_review",
            "message": "Metric not supported by current rules.",
            "evidence": []
        }

    matches = []

    for result in results:
        text = result.get("evidence", "")
        sentences = re.split(r"(?<=[.!?])\s+", text)

        for sentence in sentences:
            sentence_lower = sentence.lower().strip()

            if any(
                phrase in sentence_lower
                for phrase in metric_rules[selected_metric]
            ):
                matches.append({
                    "pdf_page": result.get("pdf_page"),
                    "evidence": sentence.strip(),
                    "source_url": result.get("source_url")
                })

    # Remove duplicate evidence while preserving source references.
    unique_matches = []
    seen = set()

    for match in matches:
        key = (match["pdf_page"], match["evidence"])

        if key not in seen:
            seen.add(key)
            unique_matches.append(match)

    return {
        "status": (
            "evidence_found" if unique_matches else "needs_review"
        ),
        "metric": selected_metric,
        "evidence": unique_matches
    }

print("Updated get_clean_metric_evidence successfully.")
print("Supported metrics:", 13)
```

```python

import re

def get_clean_metric_evidence_v3(question, results):
    metric_aliases = {
        "current ratio": ["current ratio"],
        "debt-equity ratio": ["debt-equity ratio", "debt equity ratio"],
        "debt service coverage ratio": ["debt service coverage ratio"],
        "return on equity": ["return on equity"],
        "inventory turnover ratio": ["inventory turnover ratio"],
        "trade receivables turnover ratio": ["trade receivables turnover ratio"],
        "trade payables turnover ratio": ["trade payables turnover ratio"],
        "net capital turnover ratio": ["net capital turnover"],
        "net profit margin": ["net profit margin"],
        "return on capital employed": ["return on capital employed"],
        "return on investment": ["return on investment"],
        "interest coverage ratio": ["interest coverage ratio"],
        "long term debt to working capital": [
            "long term debt to working capital"
        ],
        "return on net worth": ["return on net worth"]
    }

    question_lower = question.lower()

    selected_metric = next(
        (
            metric for metric in sorted(
                metric_aliases, key=len, reverse=True
            )
            if metric in question_lower
            or any(
                alias in question_lower
                for alias in metric_aliases[metric]
            )
        ),
        None
    )

    if selected_metric is None:
        return {
            "status": "needs_review",
            "message": "Metric not recognized.",
            "evidence": []
        }

    # Recognize the next numbered ratio row or explanatory note.
    row_pattern = re.compile(
        r"(?<!\w)(?:\d+\s+)?"
        r"(Current Ratio|Debt-Equity Ratio|Debt Service Coverage Ratio|"
        r"Return on Equity|Inventory Turnover Ratio|"
        r"Trade Receivables Turnover Ratio|Trade Payables Turnover Ratio|"
        r"Net Capital Turnover|Net Profit Margin|"
        r"Return on Capital Employed|Return on Investment|"
        r"Interest Coverage Ratio|Return on Net Worth|"
        r"Long Term Debt to Working Capital)",
        re.I
    )

    matches = []
    aliases = metric_aliases[selected_metric]

    for result in results:
        text = re.sub(r"\s+", " ", result.get("evidence", "")).strip()
        row_matches = list(row_pattern.finditer(text))

        for index, row_match in enumerate(row_matches):
            row_name = row_match.group(1).lower()

            if not any(alias in row_name for alias in aliases):
                continue

            start = row_match.start()
            end = (
                row_matches[index + 1].start()
                if index + 1 < len(row_matches)
                else len(text)
            )

            segment = text[start:end].strip()

            # Stop before a separate explanatory note.
            segment = re.split(
                r"\s+(?=[a-z]\)\s)",
                segment,
                maxsplit=1,
                flags=re.I
            )[0].strip()

            if segment:
                matches.append({
                    "pdf_page": result.get("pdf_page"),
                    "evidence": segment[:700],
                    "source_url": result.get("source_url")
                })

    unique_matches = []
    seen = set()

    for match in matches:
        key = (match["pdf_page"], match["evidence"])
        if key not in seen:
            seen.add(key)
            unique_matches.append(match)

    return {
        "status": "evidence_found" if unique_matches else "needs_review",
        "metric": selected_metric,
        "evidence": unique_matches
    }

print("Created get_clean_metric_evidence_v3.")
```

```python

import re

def run_finance_research(question):
    search_results = research_search_v2(question, max_results=5)

    evidence_result = get_clean_metric_evidence_v3(
        question, search_results
    )

    if evidence_result["status"] != "evidence_found":
        return {
            "question": question,
            "status": "needs_review",
            "answer": "Insufficient metric-specific evidence. Manual review required.",
            "sources": []
        }

    metric = evidence_result["metric"]
    evidence = evidence_result["evidence"]

    # Reported values for metrics confirmed in the extracted ratio table.
    metric_values = {
        "current ratio": ("0.99", "1.05", "(5.7)"),
        "debt-equity ratio": ("0.41", "0.37", "10.8"),
        "return on equity": ("8.5%", "7.3%", "17.0"),
        "inventory turnover ratio": ("5.91", "6.85", "(13.7)"),
        "return on investment": ("9.5%", "10.3%", "(7.7)"),
        "net profit margin": ("8.0%", "6.3%", "27.0")
    }

    # Find a row that contains the expected values for the selected metric.
    value_row = None

    if metric in metric_values:
        expected = metric_values[metric]

        for item in evidence:
            text = item.get("evidence", "")
            if all(value in text for value in expected):
                value_row = item
                break

    # Original metrics: preserve their previously tested explanations.
    if metric == "net profit margin":
        answer = (
            "Reliance Industries Limited's standalone net profit margin "
            "increased from 6.3% in FY 2024-25 to 8.0% in FY 2025-26, "
            "an increase of 1.7 percentage points. The annual report "
            "attributes the improvement to higher net profit."
        )

    elif metric == "debt service coverage ratio":
        answer = (
            "The annual report states that the debt service coverage "
            "ratio increased due to lower finance costs and principal "
            "repayments of loans during the year."
        )

    elif metric == "return on capital employed":
        answer = (
            "The annual report attributes the increase in return on "
            "capital employed to higher profit and a reduction in "
            "capital employed."
        )

    elif value_row:
        current, previous, change = metric_values[metric]
        answer = (
            f"The annual report's standalone ratio analysis reports "
            f"{metric} at {current} for FY 2025-26, compared with "
            f"{previous} for FY 2024-25. The report's change column "
            f"shows {change}%. Refer to the cited evidence for the "
            f"reported figures."
        )

    else:
        return {
            "question": question,
            "status": "needs_review",
            "answer": (
                "Evidence was retrieved, but the expected value row "
                "could not be confirmed. Manual review is required."
            ),
            "sources": evidence
        }

    # Return only the matching value row when one is available.
    sources = [value_row] if value_row else evidence

    return {
        "question": question,
        "status": "evidence_found",
        "answer": answer,
        "sources": sources
    }

print("Updated run_finance_research successfully.")
```

Build an evidence quality check

```python

from pathlib import Path
import pandas as pd
import re

OUTPUT_DIR = Path("/content/finance_research_output")
evidence_path = OUTPUT_DIR / "financial_metric_evidence_v2.csv"

evidence = pd.read_csv(evidence_path)

# Terms that should appear in evidence text for each metric.
METRIC_TERMS = {
    "Net Profit Margin": ["net profit margin"],
    "Debt Service Coverage Ratio": ["debt service coverage ratio"],
    "Return on Capital Employed": ["return on capital employed"],
    "Long Term Debt to Working Capital": [
        "long term debt to working capital"
    ],
    "Current Ratio": ["current ratio"],
    "Debt-Equity Ratio": ["debt-equity ratio", "debt equity ratio"],
    "Return on Equity": ["return on equity"],
    "Inventory Turnover Ratio": ["inventory turnover ratio"],
    "Trade Receivables Turnover Ratio": [
        "trade receivables turnover ratio"
    ],
    "Trade Payables Turnover Ratio": [
        "trade payables turnover ratio"
    ],
    "Return on Investment": ["return on investment"],
}

def check_evidence(row):
    metric = str(row.get("metric", "")).strip()
    text = str(row.get("evidence_text", "")).lower()
    expected_terms = METRIC_TERMS.get(metric)

    if not text or text == "nan":
        return "REVIEW: empty evidence"

    if expected_terms and not any(term in text for term in expected_terms):
        return "REVIEW: metric name not found in evidence text"

    if len(text.split()) < 8:
        return "REVIEW: evidence excerpt may be too short"

    return "CHECK: metric term found; verify figures and source manually"

evidence["quality_check"] = evidence.apply(check_evidence, axis=1)

quality_path = OUTPUT_DIR / "evidence_quality_check.csv"
evidence.to_csv(quality_path, index=False)

print("Evidence quality check saved:", quality_path)
print("\nResults by status:")
print(evidence["quality_check"].value_counts().to_string())

print("\nRows requiring review:")
review_rows = evidence[
    evidence["quality_check"].str.startswith("REVIEW")
]
display(
    review_rows[
        ["metric", "pdf_page", "evidence_text", "quality_check"]
    ].head(10)
)

print("\nDAY 3 — TASK 3 COMPLETE")
```

DAY 4 :- Validate the financial ratios

```python

from pathlib import Path
import pandas as pd

OUTPUT_DIR = Path("/content/finance_research_output")

ratio_path = OUTPUT_DIR / "validated_ratio_candidates.csv"
result_path = OUTPUT_DIR / "ratio_validation_results.csv"
audit_path = OUTPUT_DIR / "ratio_manual_audit.csv"

ratios = pd.read_csv(ratio_path)

# Reference figures transcribed from the ratio table in the
# existing RIL annual-report research notes.
# These are comparison targets, not independent verification.
reference_values = {
    "Current Ratio": (0.99, 1.05, "ratio"),
    "Debt-Equity Ratio": (0.41, 0.37, "ratio"),
    "Debt Service Coverage Ratio": (4.03, 2.06, "ratio"),
    "Return on Equity": (8.5, 7.3, "%"),
    "Net Profit Margin": (8.0, 6.3, "%"),
    "Return on Capital Employed": (20.7, 14.6, "%"),
    "Return on Investment": (9.5, 10.3, "%"),
}

audit_rows = []

for _, row in ratios.iterrows():
    metric = str(row.get("metric", "")).strip()
    expected = reference_values.get(metric)

    current = pd.to_numeric(
        pd.Series([row.get("fy_2025_26")]), errors="coerce"
    ).iloc[0]
    previous = pd.to_numeric(
        pd.Series([row.get("fy_2024_25")]), errors="coerce"
    ).iloc[0]

    if expected is None:
        audit_rows.append({
            "metric": metric,
            "candidate_fy_2025_26": current,
            "reference_fy_2025_26": None,
            "candidate_fy_2024_25": previous,
            "reference_fy_2024_25": None,
            "comparison": "REVIEW: no reference value configured",
            "source_check": "Manual source verification required"
        })
        continue

    ref_current, ref_previous, ref_unit = expected

    current_match = pd.notna(current) and abs(current - ref_current) < 0.011
    previous_match = pd.notna(previous) and abs(previous - ref_previous) < 0.011

    audit_rows.append({
        "metric": metric,
        "candidate_fy_2025_26": current,
        "reference_fy_2025_26": ref_current,
        "candidate_fy_2024_25": previous,
        "reference_fy_2024_25": ref_previous,
        "comparison": (
            "MATCHES REFERENCE"
            if current_match and previous_match
            else "DISCREPANCY: inspect values"
        ),
        "source_check": "Still requires manual verification",
        "unit": ref_unit,
        "pdf_page": row.get("pdf_page", "")
    })

audit = pd.DataFrame(audit_rows)
audit.to_csv(audit_path, index=False)

print("DAY 4 — TASK 1 COMPLETE")
print("Audit file saved:", audit_path)
print("\nComparison results:")
display(audit)
print("\nReminder: matching the reference values is not independent source verification.")
```

Calculate year-on-year changes and strengthen validation

```python

from pathlib import Path
import pandas as pd
import numpy as np

OUTPUT_DIR = Path("/content/finance_research_output")

ratio_path = OUTPUT_DIR / "validated_ratio_candidates.csv"
audit_path = OUTPUT_DIR / "ratio_manual_audit.csv"
output_path = OUTPUT_DIR / "ratio_change_analysis.csv"

ratios = pd.read_csv(ratio_path)
audit = pd.read_csv(audit_path)

rows = []

for _, row in ratios.iterrows():
    metric = str(row["metric"])
    current = pd.to_numeric(
        pd.Series([row["fy_2025_26"]]), errors="coerce"
    ).iloc[0]
    previous = pd.to_numeric(
        pd.Series([row["fy_2024_25"]]), errors="coerce"
    ).iloc[0]

    if pd.notna(current) and pd.notna(previous) and previous != 0:
        change_pct = ((current - previous) / abs(previous)) * 100
    else:
        change_pct = np.nan

    rows.append({
        "metric": metric,
        "fy_2024_25": previous,
        "fy_2025_26": current,
        "calculated_change_pct": round(change_pct, 2)
            if pd.notna(change_pct) else np.nan,
        "interpretation_note": (
            "Increase" if current > previous else
            "Decrease" if current < previous else
            "No change" if current == previous else
            "Cannot calculate"
        ),
        "source_verification": "Pending original annual-report check"
    })

changes = pd.DataFrame(rows)

# Join the comparison status from the previous task.
changes = changes.merge(
    audit[["metric", "comparison", "source_check"]],
    on="metric",
    how="left"
)

changes.to_csv(output_path, index=False)

print("DAY 4 — TASK 2 COMPLETE")
print("Saved:", output_path)
display(changes)
```

```python

import json
from pathlib import Path

path = Path("/content/finance_research_output/ril_annual_report_pages.json")

with open(path, "r", encoding="utf-8") as f:
    data = json.load(f)

print("Data type:", type(data).__name__)

if isinstance(data, dict):
    print("Top-level keys:", list(data.keys())[:10])
    pages = data.get("pages", data)
elif isinstance(data, list):
    pages = data
else:
    raise ValueError("Unexpected JSON structure")

def show_page(page_number):
    # Support either a list of page records or a dictionary keyed by page number.
    if isinstance(pages, list):
        matches = [
            p for p in pages
            if isinstance(p, dict)
            and p.get("page_number", p.get("page")) in
                (page_number, page_number - 1)
        ]
        if matches:
            item = matches[0]
        elif 0 <= page_number - 1 < len(pages):
            item = pages[page_number - 1]
        else:
            print("Page not found:", page_number)
            return
    elif isinstance(pages, dict):
        item = pages.get(str(page_number),
                         pages.get(str(page_number - 1)))
        if item is None:
            print("Page not found:", page_number)
            return
    else:
        print("Unexpected pages structure:", type(pages).__name__)
        return

    print(f"\n{'=' * 25} PAGE {page_number} {'=' * 25}")
    if isinstance(item, dict):
        print(item.get("text", item.get("content", str(item)))[:7000])
    else:
        print(str(item)[:7000])

for page_number in [6, 93]:
    show_page(page_number)
```

Generate the final report version

```python

from pathlib import Path
import pandas as pd

base = Path("/content/finance_research_output")
report_path = base / "reports" / "RIL_Financial_Research_Draft.md"
evidence_path = base / "financial_evidence_register_v2.csv"
explanations_path = base / "financial_evidence_explanations_v2.csv"
audit_path = base / "ratio_change_audit_v2.csv"

for p in [report_path, evidence_path, explanations_path, audit_path]:
    if not p.exists():
        raise FileNotFoundError(f"Required file missing: {p}")

draft = report_path.read_text(encoding="utf-8")
evidence = pd.read_csv(evidence_path).fillna("")
explanations = pd.read_csv(explanations_path).fillna("")
audit = pd.read_csv(audit_path).fillna("")

def md_table(df, columns, headers=None):
    view = df[columns].copy()
    if headers:
        view.columns = headers
    view = view.astype(str).replace({"nan": "", "None": ""})
    return view.to_markdown(index=False)

appendix = []

appendix.append("\n\n---\n\n# Final Evidence and Validation Appendix\n")
appendix.append(
    "\n## A. Financial-ratio evidence register\n\n"
    "Values below are transcribed from the extracted annual report, "
    "Standalone Financial Statements, Note 40, Ratio Analysis "
    "(extracted PDF page 93). Visual verification against the original PDF "
    "remains a final review step.\n\n"
)
appendix.append(md_table(
    evidence,
    ["metric", "fy_2025_26", "fy_2024_25",
     "company_reported_change_pct", "unit", "pdf_page"],
    ["Metric", "FY 2025–26", "FY 2024–25",
     "Company-reported change (%)", "Unit", "PDF page"]
))

appendix.append(
    "\n\n## B. Reported changes versus calculated changes\n\n"
    "Calculated changes use the displayed values in the ratio table. "
    "Differences can arise when the company calculates changes from "
    "underlying values with more precision than the rounded figures shown.\n\n"
)
if "company_reported_change_pct" in audit.columns:
    appendix.append(md_table(
        audit,
        ["metric", "company_reported_change_pct", "calculated_change_pct",
         "change_difference_percentage_points", "review_note"],
        ["Metric", "Reported change (%)", "Calculated change (%)",
         "Difference (percentage points)", "Review note"]
    ))
else:
    appendix.append(
        "The ratio-change audit file did not contain the expected columns; "
        "see ratio_change_audit_v2.csv for the detailed audit.\n"
    )

appendix.append("\n\n## C. Explanations stated in the annual report\n\n")
for _, row in explanations.iterrows():
    appendix.append(
        f"### {row['metric']}\n\n"
        f"{row['source_explanation']} "
        f"(extracted PDF page {row['pdf_page']}).\n\n"
    )

appendix.append(
    "## D. Validation status and limitations\n\n"
    "- The ratio values above were transcribed from extracted annual-report text.\n"
    "- The original PDF table should be visually checked before claiming "
    "independent source verification.\n"
    "- The company-reported percentage changes and calculations based on "
    "displayed rounded values are presented separately.\n"
    "- ROE and ROCE show differences between the reported change and the "
    "calculation from displayed values; underlying unrounded figures have "
    "not been confirmed here.\n"
    "- Net Capital Turnover Ratio is shown as unavailable for FY 2025–26 "
    "in the source table; the report notes negative working capital.\n"
    "- Extracted text and automated evidence matching may contain errors. "
    "Findings should be checked against the cited source.\n"
    "- This report is for educational and research purposes and is not "
    "personalized investment advice.\n"
)

final_path = base / "reports" / "RIL_Financial_Research_Final.md"
final_path.write_text(draft + "".join(appendix), encoding="utf-8")

print("Created:", final_path)
print("Original draft preserved:", report_path.exists())
print("Final report characters:", len(final_path.read_text(encoding="utf-8")))
print("\nFinal report ending:")
print(final_path.read_text(encoding="utf-8")[-1800:])
```

```python
# Day 6 — Targeted consolidated revenue fix
# Preserve the existing, tested research function.
_run_finance_research_existing = run_finance_research

def run_finance_research(question):
    q = question.lower()

    is_revenue_question = (
        "revenue" in q
        and not any(term in q for term in [
            "retail revenue", "segment revenue",
            "digital services revenue", "o2c revenue"
        ])
    )

    if not is_revenue_question:
        return _run_finance_research_existing(question)

    # Search existing extracted annual-report pages for
    # the consolidated revenue figures and their labels.
    sources = []

    for page in report_pages:
        text = re.sub(r"\s+", " ", page.get("text", "")).strip()

        if (
            "11,75,919" in text
            and "10,71,174" in text
            and (
                "value of sales" in text.lower()
                or "consolidated revenues" in text.lower()
            )
        ):
            # Prefer the financial statements with the explicit
            # revenue label and comparison-year figures.
            lower = text.lower()
            pos = lower.find("value of sales")
            if pos == -1:
                pos = lower.find("consolidated revenues")

            snippet = text[max(0, pos - 150):pos + 650]

            sources.append({
                "pdf_page": page.get("pdf_page"),
                "evidence": snippet,
                "source_url": page.get("source_url")
            })

    if not sources:
        return {
            "question": question,
            "status": "needs_review",
            "answer": "Consolidated revenue evidence could not be located.",
            "sources": []
        }

    # Prefer a source that explicitly labels the financial statement.
    preferred = next(
        (
            s for s in sources
            if "value of sales" in s["evidence"].lower()
            and "revenue from operations" in s["evidence"].lower()
        ),
        sources[0]
    )

    answer = (
        "Reliance Industries Limited's consolidated Value of Sales "
        "and Services (Revenue) was ₹11,75,919 crore in FY 2025-26, "
        "compared with ₹10,71,174 crore in FY 2024-25. The annual "
        "report separately reports Revenue from Operations, net of "
        "GST, of ₹10,75,675 crore for FY 2025-26 and ₹9,80,136 crore "
        "for FY 2024-25. These are distinct reported measures."
    )

    return {
        "question": question,
        "status": "evidence_found",
        "answer": answer,
        "sources": [preferred]
    }

print("Targeted revenue handler installed.")
```
